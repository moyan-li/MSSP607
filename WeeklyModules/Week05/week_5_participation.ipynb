```python
import pandas as pd

aapl = pd.read_csv("/content/MSSP607/WeeklyModules/Week05/aapl.csv")
amzn = pd.read_csv("/content/MSSP607/WeeklyModules/Week05/amzn.csv")
fb = pd.read_csv("/content/MSSP607/WeeklyModules/Week05/fb.csv")
goog = pd.read_csv("/content/MSSP607/WeeklyModules/Week05/goog.csv")
nflx = pd.read_csv("/content/MSSP607/WeeklyModules/Week05/nflx.csv")
```

```python
aapl["ticker"] = "AAPL"
amzn["ticker"] = "AMZN"
fb["ticker"] = "FB"
goog["ticker"] = "GOOG"
nflx["ticker"] = "NFLX"
```

```python
faang = pd.concat(
    [fb, aapl, amzn, nflx, goog],
    ignore_index=True
)

faang.head()
```

```python
faang.to_csv("faang.csv", index=False)
```

```python
faang = pd.read_csv("faang.csv")

faang["date"] = pd.to_datetime(faang["date"])
faang["volume"] = faang["volume"].astype(int)

faang = faang.sort_values(
    by=["date", "ticker"]
)

faang.head()
```

```python
faang.nlargest(7, "volume")
```

```python
faang_melted = faang.melt(
    id_vars=["date", "ticker"]
)

faang_melted.head()
```
