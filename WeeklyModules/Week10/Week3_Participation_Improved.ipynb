## Questions about functions and scope

**a. When does code inside a function execute?**  
When the function is called, not when it is defined.

**b. How many global and local scopes are there?**  
Each Python module has its own global scope. Each function call creates a new local scope. Multiple calls can have separate local scopes at the same time.

**c. What happens to local variables when a function returns?**  
They are normally no longer accessible through that function call. Objects can remain available if they are returned or referenced elsewhere; they are not necessarily destroyed immediately.

**d. How can you handle an error without stopping the program?**  
Put code that may fail inside a `try` block and catch the specific expected exception with `except`. For example, catch `ValueError` when converting invalid text to an integer.

## Collatz sequence

For a positive integer, divide it by 2 if it is even; otherwise multiply it by 3 and add 1. Repeat until the value reaches 1. The sequence includes the starting number. For example, input `3` produces `3 -> 10 -> 5 -> 16 -> 8 -> 4 -> 2 -> 1`.

Whether every positive integer eventually reaches 1 is an unproven mathematical conjecture. The loop stops when it reaches 1; an unusually long run can be interrupted in the notebook.

Run the function cell below, then the interactive cell. No Google Drive mount, GitHub token, or external packages are needed for this exercise.


```python
def collatz(number):
    """Return the next Collatz number for a positive integer."""
    if not isinstance(number, int):
        raise TypeError("number must be an integer.")
    if number <= 0:
        raise ValueError("number must be greater than 0.")

    if number % 2 == 0:
        return number // 2
    return 3 * number + 1


def read_positive_integer():
    """Keep asking until the user enters a positive integer."""
    while True:
        response = input("Enter a starting integer greater than 0: ").strip()
        try:
            number = int(response)
        except ValueError:
            print("Invalid input. Please enter a whole number, such as 3.")
            continue

        if number > 0:
            return number
        print("The number must be greater than 0. Please try again.")
```

## Run the program
Enter a positive integer when prompted. Invalid input will prompt you to try again. Input `1` prints `1` without extra steps.


```python
number = read_positive_integer()
print(number, end="")

while number != 1:
    number = collatz(number)
    print(f" -> {number}", end="")

print()  # Finish the output with a newline.
```
