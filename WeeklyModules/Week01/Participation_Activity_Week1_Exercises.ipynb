## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
sum(salaries)/len(salaries)
print(sum(salaries)/len(salaries))
import statistics
print(statistics.mean(salaries))
```

### median

```python
sorted_salaries = sorted(salaries)
if len(sorted_salaries) % 2 == 0:
    median = (sorted_salaries[len(salaries) // 2] + sorted_salaries[len(sorted_salaries) // 2 -1]) / 2
else:
    median = sorted_salaries[len(salaries) // 2]
print(median)
print(statistics.median(sorted_salaries))
```

### mode

```python
count_dict = {}
for num in salaries:
    if num in count_dict:
        count_dict[num] += 1
    else:
        count_dict[num] = 1
max_count = max(count_dict.values())
mode = [key for key, value in count_dict.items() if value == max_count]
print(mode)
print(statistics.mode(salaries))
```

### sample variance
Remember to use Bessel's correction.

```python
salaries_var = sum([(salary - statistics.mean(salaries)) ** 2 for salary in salaries]) / (len(salaries) - 1)
print(salaries_var)
print(statistics.variance(salaries))
```

### sample standard deviation
Remember to use Bessel's correction.

```python
salaries_std = salaries_var ** 0.5
print(salaries_std)
print(statistics.stdev(salaries))
```

## Exercise 6: Calculating more statistics
### range

```python
range = max(salaries) - min(salaries)
print(range)
```

### coefficient of variation

```python
salary_cv = salaries_std / statistics.mean(salaries)
print(salary_cv)
```

### interquartile range

```python

low=sorted_salaries[:len(sorted_salaries) // 2]
high=sorted_salaries[len(sorted_salaries) // 2:]
if len(low) % 2 == 0:
    Q1 = (low[len(low) // 2] + low[len(low) // 2 -1]) / 2
else:
    Q1 = low[len(low) // 2]
if len(high) % 2 == 0:
    Q3 = (high[len(high) // 2] + high[len(high) // 2 -1]) / 2
else:
    Q3 = high[len(high) // 2]
print(Q3 - Q1)
q1 = statistics.median(sorted_salaries[:len(sorted_salaries) // 2])
q3 = statistics.median(sorted_salaries[len(sorted_salaries) // 2:])
iqr = q3 - q1
print(iqr)
```

### quartile coefficent of dispersion

```python
qcd= iqr / (q3 + q1)
print(qcd)
```

## Exercise 7: Scaling data
### min-max scaling

```python
scaled_salaries = [(salary - min(salaries)) / (max(salaries) - min(salaries)) for salary in salaries]
print(scaled_salaries)
```

### standardizing

```python
stand_salaries=[(salary - statistics.mean(salaries)) / statistics.stdev(salaries)for salary in salaries]
print(stand_salaries)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
statistics.covariance(stand_salaries, scaled_salaries)
```

### Pearson correlation coefficient ($\rho$)

```python
statistics.correlation(stand_salaries, scaled_salaries)
```
