```python
letters = ['a', 'b', 'c', 'd']
print(letters[int(int('3' * 2) // 11)])
# '3' * 2 = '33'；int('33') is 33； 33 // 11 is 3；letters[3] is 'd'

print(letters[-1]) # the last one
print(letters[:2])
# Before fetching index 2 from the beginning, index 2 is not included, so the first two elements are obtained
```

```python
spam = ['apples', 'bananas', 'tofu', 'cats']
def code1(thelist):
   if len(thelist) == 0:
      return ''
   elif len(thelist) == 1:
      return thelist[0]
   else:
      return ', '.join(spam[:-1])
print(code1(spam) + ', and ' + spam[-1])
```
