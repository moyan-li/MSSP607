```python
import pandas as pd
import numpy  as np
df = pd.read_csv('/content/MSSP607/WeeklyModules/Week04/parsed.csv')
```

```python
df.columns
```

```python
#a.Find the 95th percentile of earthquake magnitude in Japan using magType of 'mb'
location = df["place"].fillna("").astype(str)
japan_mb_95 = df.loc[
    (location.str.contains("Japan", case=False, na=False)) &
    (df["magType"].eq("mb")),
    "mag"
].quantile(0.95)

print(japan_mb_95)
```

```python
#b. Find the percentage of earthquakes in Indonesia that were coupled with tsunamis.
indonesia = df.loc[
    location.str.contains("Indonesia", case=False, na=False)
]

indonesia_tsunami_pct = indonesia["tsunami"].eq(1).mean() * 100

print(f"{indonesia_tsunami_pct:.2f}%")
```

```python
#c. Get summary statistics for earthquakes in Nevada
nevada = df.loc[
    location.str.contains("Nevada", case=False, na=False)
]
print(nevada.describe())
```

```python
#d. Add a column to the dataframe indicating whether or not the earthquake happened in a country or US state that is on the Ring of Fire.

ring_pattern = (
"Taiwan"
)
df["ring_of_fire"] = location.str.contains(
    ring_pattern,
    case=False,
    regex=True,
    na=False
)

df[["place", "ring_of_fire"]].head()
```

```python
#e.Calculate the number of earthquakes in the Ring of Fire locations and the number outside of them
ring_counts = df["ring_of_fire"].value_counts()

print("in the Ring of Fire：", ring_counts.get(True, 0))
print("outside of Ring of Fire：", ring_counts.get(False, 0))
```

```python
#f. Find the tsunami count along the Ring of Fire.
ring_tsunami_count = df.loc[
    (df["ring_of_fire"]) &
    (df["tsunami"].eq(1))
].shape[0]

print(ring_tsunami_count)
```
